```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# 1. GA Convergence Plot
# -------------------------------
gens = np.arange(1, 101)
best_fitness = 1000 / (1 + 0.05 * gens)  # synthetic decreasing curve
avg_fitness = best_fitness * 1.2

plt.figure(figsize=(8,5))
plt.plot(gens, best_fitness, label="Best Fitness", color="blue")
plt.plot(gens, avg_fitness, label="Average Fitness", color="orange", linestyle="--")
plt.xlabel("Generations")
plt.ylabel("Fitness")
plt.title("GA Convergence: Best and Average Fitness")
plt.legend()
plt.grid(True)
plt.show()

# -------------------------------
# 2. Patient Arrival Pattern
# -------------------------------
hours = np.arange(168)
# synthetic arrivals: baseline + surge + weekend
arrivals = 10 + 5*np.sin(2*np.pi*hours/24)  # diurnal
arrivals[48:120] *= 1.8  # surge days 3–5
arrivals[120:] *= 0.7    # weekend

plt.figure(figsize=(10,4))
plt.plot(hours, arrivals, color="green")
plt.xlabel("Hour")
plt.ylabel("Patient Arrivals")
plt.title("Synthetic Patient Arrival Pattern (7 days)")
plt.grid(True)
plt.show()

# -------------------------------
# 3. Resource Allocation Comparison
# -------------------------------
labels = ["Nurses","Doctors","Technicians","ICU Beds","General Beds"]
baseline = [30,15,10,25,85]
optimized = [32,16,10,24,74]

x = np.arange(len(labels))
width = 0.35

plt.figure(figsize=(8,5))
plt.bar(x - width/2, baseline, width, label="Baseline")
plt.bar(x + width/2, optimized, width, label="Optimized")
plt.xticks(x, labels)
plt.ylabel("Average Allocation")
plt.title("Baseline vs Optimized Resource Allocation")
plt.legend()
plt.show()

# -------------------------------
# 4. Sensitivity Analysis (Weights)
# -------------------------------
scenarios = ["Cost-focused","Balanced","Quality-focused"]
wait_times = [2.18,1.72,1.21]
costs = [33200,39450,47800]

plt.figure(figsize=(8,5))
plt.plot(scenarios, wait_times, marker="o", label="Wait Time (hrs)")
plt.plot(scenarios, costs, marker="s", label="Cost ($)")
plt.title("Sensitivity to Objective Weights")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
```
